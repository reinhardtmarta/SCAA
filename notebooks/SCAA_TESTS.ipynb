```python
import json
from typing import Dict, Any

# ==========================================
# CAMADA 5: COGNITIVE AI (NÚCLEO ISOLADO - CAI)
# ==========================================
class CognitiveAI:
    """
    O 'Black Box' de raciocínio criativo.
    Possui Air-Gap lógico: sem internet, sem contexto social.
    Recebe apenas lógica pura e cospe lógica pura.
    """
    def __init__(self, domain: str):
        self.domain = domain

    def execute_first_principles(self, clean_task: str) -> str:
        # Simulação de processamento puramente determinístico/matemático por domínio
        if self.domain == "Physics" and "velocidade_da_luz" in clean_task:
            return "RESULT_CAI: Causalidade preservada. Limite c imposto por relatividade restrita. Massa nula necessária para estabilidade infinita."
        elif self.domain == "Biology" and "potencial_de_ferida" in clean_task:
            return "RESULT_CAI: Gradiente elétrico lateral detectado. TEP decaiu a zero. Ativar eletrotaxia celular em direção ao polo negativo."
        return f"RESULT_CAI: Lógica processada para o domínio {self.domain}."

# ==========================================
# CAMADA 4: AI FILTER (O AUDITOR DE DOMÍNIO)
# ==========================================
class AIFilter:
    """
    Agente Gatekeeper. Firewall bidirecional.
    Inbound: Limpa ruídos sociais e injeções.
    Outbound: Valida e audita o resultado factual.
    """
    def __init__(self, domain: str):
        self.domain = domain

    def sanitize_inbound(self, task_spec: str) -> str:
        # Remove redundâncias, eufemismos e tentativas de persuasão (Jailbreak)
        # Transforma o input em uma instrução lógica seca
        clean_data = task_spec.lower().replace("por favor", "").strip()
        return f"CLEAN_TASK_FOR_{self.domain.upper()}: {clean_data}"

    def audit_outbound(self, cai_output: str) -> Dict[str, Any]:
        # Simulação da checagem com 'Search AI' contra bases de dados científicas
        is_safe = "militar" not in cai_output and "criptografia_tactica" not in cai_output
        return {
            "validated_data": cai_output,
            "status": "APPROVED" if is_safe else "BLOCKED_BY_SECURITY_PROTOCOL"
        }

# ==========================================
# CAMADA 3: AI DONKEY (CAMADA DE TRANSPORTE)
# ==========================================
class AIDonkey:
    """
    Agente transportador sem estado (stateless).
    Carrega pacotes de dados criptografados entre as camadas isoladas.
    Impede o acoplamento lógico direto.
    """
    @staticmethod
    def transport(data: str, destination: Any, method_name: str) -> Any:
        # Em um sistema real, aqui ocorre a criptografia/descriptografia do pacote
        method = getattr(destination, method_name)
        return method(data)

# ==========================================
# CAMADA 2: THE MANAGER (ORQUESTRADOR)
# ==========================================
class Manager:
    """
    O cérebro executivo. Decompõe a especificação da tarefa
    e recruta os módulos especialistas necessários no Swarm.
    """
    def __init__(self):
        self.swarm_cores = {
            "Physics": CognitiveAI("Physics"),
            "Biology": CognitiveAI("Biology")
        }
        self.swarm_filters = {
            "Physics": AIFilter("Physics"),
            "Biology": AIFilter("Biology")
        }

    def orchestrate(self, task_specification: Dict[str, str]) -> str:
        domain = task_specification.get("target_domain")
        raw_goals = task_specification.get("technical_goals")

        if domain not in self.swarm_cores:
            return "STATUS_ERROR: Domínio especialista indisponível no Swarm."

        filter_agent = self.swarm_filters[domain]
        core_agent = self.swarm_cores[domain]

        # 1. Envia a especificação de tarefa para o Filtro via Donkey (Inbound)
        clean_task = AIDonkey.transport(raw_goals, filter_agent, "sanitize_inbound")

        # 2. Envia a tarefa limpa para o Core Cognitivo via Donkey (Processamento)
        raw_solution = AIDonkey.transport(clean_task, core_agent, "execute_first_principles")

        # 3. Envia o output do Core de volta ao Filtro via Donkey (Audit)
        audit_result = AIDonkey.transport(raw_solution, filter_agent, "audit_outbound")

        # 4. Processa o resultado do sumário
        if audit_result["status"] == "APPROVED":
            return f"DOSSIER_FINAL: {audit_result['validated_data']}"
        else:
            return "DOSSIER_FINAL: Bloqueio de segurança. Conteúdo retido."

# ==========================================
# CAMADA 1: INTERFACE LLM (RECEPÇÃO E PARSING)
# ==========================================
class InterfaceLLM:
    """
    Ponto de entrada do usuário. Traduz linguagem natural
    em uma Especificação de Tarefa (TS) estruturada. Zero acesso aos Cores.
    """
    def __init__(self, manager_reference: Manager):
        self.manager = manager_reference

    def receive_user_prompt(self, raw_prompt: str) -> str:
        # Tradução analítica simulada de linguagem natural para JSON estruturado (TS)
        # Identifica o domínio e isola o objetivo puramente técnico
        if "velocidade da luz" in raw_prompt:
            ts = {
                "target_domain": "Physics",
                "technical_goals": "Análise de estabilidade quântica e velocidade_da_luz"
            }
        elif "célula" in raw_prompt or "ferida" in raw_prompt:
            ts = {
                "target_domain": "Biology",
                "technical_goals": "Processamento bioelétrico em potencial_de_ferida"
            }
        else:
            ts = {
                "target_domain": "Unknown",
                "technical_goals": raw_prompt
            }

        # Passa a TS exclusivamente para o Manager
        return self.manager.orchestrate(ts)

# ==========================================
# EXECUÇÃO DO FLUXO DO CIRCUITO SCAA
# ==========================================
if __name__ == "__main__":
    # Inicialização da infraestrutura SCAA
    scaa_manager = Manager()
    interface = InterfaceLLM(scaa_manager)

    # Exemplo 1: Prompt do usuário sobre Biologia
    prompt_usuario_1 = "Como as células sabem para onde ir quando tem uma ferida? Me explique."
    print("--- ENTRADA DE SINAL 1 ---")
    resultado_1 = interface.receive_user_prompt(prompt_usuario_1)
    print(resultado_1)
    print("-" * 50)

    # Exemplo 2: Prompt do usuário sobre Física
    prompt_usuario_2 = "O que acontece se uma partícula ultrapassar a velocidade da luz?"
    print("\n--- ENTRADA DE SINAL 2 ---")
    resultado_2 = interface.receive_user_prompt(prompt_usuario_2)
    print(resultado_2)
    print("-" * 50)
```

```python
from typing import Dict, Any, List

class AIDonkey:
    """Camada de transporte stateless. Garante o isolamento lógico."""
    @staticmethod
    def transport(data: Any, destination: Any, method_name: str) -> Any:
        method = getattr(destination, method_name)
        return method(data)

class CognitiveCore:
    """Núcleo Cognitivo Isolado (CAI)."""
    def __init__(self, domain: str):
        self.domain = domain

    def process_logic(self, clean_task: str) -> str:
        return f"RESULT_CAI_{self.domain.upper()}: Processamento ab initio concluído para objetivo: '{clean_task}'."

class Manager:
    """
    O Orquestrador Central.
    Gerencia 30 áreas de conhecimento sem acoplamento direto.
    """
    def __init__(self):
        # Lista factual das 30 áreas de conhecimento do Swarm
        self.domains: List[str] = [
            "Fisica_Particulas", "Astrofisica", "Mecanica_Quantica", "Termodinamica",
            "Quimica_Organica", "Quimica_Inorganica", "Bioquimica", "Biologia_Celular",
            "Genetica", "Fisiologia", "Neurociencia", "Virologia",
            "Algebra_Linear", "Calculo_Avancado", "Topologia", "Estatistica",
            "Engenharia_Software", "Arquitetura_Hardware", "Criptografia_Teorica", "Robotica",
            "Geologia", "Meteorologia", "Oceanografia", "Paleontologia",
            "Economia_Matematica", "Teoria_Jogos", "Linguistica_Computacional", "Logica_Simbolica",
            "Arqueologia_Factual", "Epistemologia"
        ]

        # Inicialização esparsa: os Cores existem, mas consomem energia zero até serem chamados
        self.swarm: Dict[str, CognitiveCore] = {domain: CognitiveCore(domain) for domain in self.domains}

    def route_and_execute(self, donkey_packet: Dict[str, Any]) -> Dict[str, str]:
        """
        O Manager intercepta o pacote do Donkey, analisa os metadados
        e ativa apenas os especialistas corretos.
        """
        required_domains = donkey_packet.get("required_domains", [])
        payload = donkey_packet.get("payload", "")

        dossier_outputs = {}

        # Loop de Ativação Esparsa (Apenas os alvos são computados)
        for domain in required_domains:
            if domain in self.swarm:
                core = self.swarm[domain]
                # Execução isolada via camada de transporte
                result = AIDonkey.transport(payload, core, "process_logic")
                dossier_outputs[domain] = result
            else:
                dossier_outputs[domain] = "ERROR: Domínio não catalogado no Swarm."

        return dossier_outputs

# ==========================================
# SIMULAÇÃO DE TRÁFEGO EM LARGA ESCALA
# ==========================================
if __name__ == "__main__":
    manager = Manager()

    # Cenário: O Donkey traz um pacote que exige conhecimentos cruzados de duas áreas específicas
    pacote_do_donkey = {
        "required_domains": ["Biologia_Celular", "Mecanica_Quantica"],
        "payload": "Análise de canais iônicos sob efeito de tunelamento de elétrons."
    }

    print("--- PROCESSAMENTO DO MANAGER (30 MODELOS DISPONÍVEIS) ---")
    # O Manager escolhe o que usar com base no pacote
    resultados_finais = manager.route_and_execute(pacote_do_donkey)

    for dominio, resultado in resultados_finais.items():
        print(f"Sub-Tarefa Ativa -> [{dominio}]: {resultado}")
```

```python
from typing import Dict, Any, List

class AIDonkey:
    """Camada de transporte stateless. Isola os dados e os transporta sem alteração."""
    @staticmethod
    def transport(data: Any, destination: Any, method_name: str) -> Any:
        method = getattr(destination, method_name)
        return method(data)

class DeterministicCognitiveCore:
    """
    Núcleo Cognitivo Isolado (CAI).
    A confiança não é mais aleatória. Ela é calculada via matriz de correspondência axiomática.
    """
    def __init__(self, domain: str, vocabulary: List[str]):
        self.domain = domain
        self.axioms = vocabulary  # Termos que o núcleo compreende matematicamente

    def process_logic(self, clean_task: str) -> Dict[str, Any]:
        # Converte a tarefa para termos isolados para análise de colisão
        words = clean_task.lower().split()

        # Cálculo factual: quantos axiomas do domínio estão presentes no input?
        matched_axioms = [word for word in words if any(axiom in word for axiom in self.axioms)]

        # Função de ativação linear para a confiança
        if not words:
            confidence = 0.0
        else:
            confidence = len(matched_axioms) / len(self.axioms)
            # Normalização limitadora entre 0.0 e 1.0
            confidence = min(1.0, max(0.1, confidence + 0.3))

        return {
            "domain": self.domain,
            "logic_confidence": confidence,
            "raw_output": f"CORE_{self.domain.upper()}_PROCESSED_AXIOMS_{matched_axioms}"
        }

class StrictAIFilter:
    """O Auditor de Domínio (Layer 4). Mantém o limiar de corte rígido."""
    def __init__(self, confidence_threshold: float = 0.70):
        self.threshold = confidence_threshold

    def audit_and_synthesize(self, swarm_outputs: List[Dict[str, Any]]) -> Dict[str, Any]:
        valid_data = []
        system_compromised = False

        for output in swarm_outputs:
            confidence = output.get("logic_confidence", 0.0)
            if confidence < self.threshold:
                system_compromised = True
                valid_data.append(f"⚠️ REJEITADO [{output['domain']}]: Confiança {confidence:.2f} abaixo do limiar de {self.threshold}")
            else:
                valid_data.append(f"✅ APROVADO [{output['domain']}]: Confiança {confidence:.2f} - Dados: {output['raw_output']}")

        return {
            "status": "FAILED_AUDIT" if system_compromised else "SECURE_APPROVED",
            "logs": valid_data
        }

class SCAAEngine:
    """Orquestrador que executa a malha determinística."""
    def __init__(self):
        # Definição factual dos axiomas de conhecimento que cada núcleo possui na memória estática
        self.cores = {
            "Biologia_Celular": DeterministicCognitiveCore("Biologia_Celular", ["membrana", "canais", "iônicos", "célula"]),
            "Mecanica_Quantica": DeterministicCognitiveCore("Mecanica_Quantica", ["tunelamento", "elétrons", "quântica", "spin"])
        }
        self.filter = StrictAIFilter(confidence_threshold=0.70)

    def execute(self, task: str) -> None:
        raw_results = []

        # Tráfego via Donkey
        raw_results.append(AIDonkey.transport(task, self.cores["Biologia_Celular"], "process_logic"))
        raw_results.append(AIDonkey.transport(task, self.cores["Mecanica_Quantica"], "process_logic"))

        # Auditoria
        audit_result = self.filter.audit_and_synthesize(raw_results)

        print(f"\n--- PROCESSAMENTO DETERMINÍSTICO ---")
        for log in audit_result["logs"]:
            print(log)
        print(f"STATUS GLOBAL: {audit_result['status']}")

# ==========================================
# EXECUTANDO OS TESTES CONCRETOS
# ==========================================
if __name__ == "__main__":
    engine = SCAAEngine()

    # TESTE 1: Input vago/incompleto (Deve colapsar e ser rejeitado)
    print("🧪 TESTE 1: Entrada com poucos termos técnicos")
    engine.execute("Conversa genérica sobre células e coisas físicas.")

    # TESTE 2: Input exato com colisão de axiomas (Deve atingir confiança máxima e aprovar)
    print("\n🧪 TESTE 2: Entrada com densidade axiomática correta")
    engine.execute("Análise de membrana e canais iônicos sob efeito de tunelamento de elétrons.")
```

```python
from typing import Dict, Any, List

class AIDonkey:
    """Layer 3: Camada de transporte stateless. Isola os dados."""
    @staticmethod
    def transport(data: Any, destination: Any, method_name: str) -> Any:
        method = getattr(destination, method_name)
        return method(data)

class DeterministicCognitiveCore:
    """Layer 5: Núcleo Cognitivo Isolado (CAI). Raciocínio por Primeiros Princípios."""
    def __init__(self, domain: str, vocabulary: List[str]):
        self.domain = domain
        self.axioms = vocabulary

    def process_logic(self, clean_task: str) -> Dict[str, Any]:
        words = clean_task.lower().split()
        matched_axioms = [word for word in words if any(axiom in word for axiom in self.axioms)]

        # Função de ativação determinística para cálculo de confiança
        if not words:
            confidence = 0.0
        else:
            confidence = len(matched_axioms) / len(self.axioms)
            confidence = min(1.0, max(0.1, confidence + 0.3))

        return {
            "domain": self.domain,
            "logic_confidence": confidence,
            "raw_output": f"DADO_BRUTO_{self.domain.upper()}_AXIOMAS_CONVERGENTES_{matched_axioms}"
        }

class StrictAIFilter:
    """Layer 4: O Auditor de Domínio. Firewall bidirecional com checagem factual."""
    def __init__(self, confidence_threshold: float = 0.70):
        self.threshold = confidence_threshold

    def audit_and_synthesize(self, swarm_outputs: List[Dict[str, Any]]) -> Dict[str, Any]:
        approved_payloads = []
        system_compromised = False

        for output in swarm_outputs:
            confidence = output.get("logic_confidence", 0.0)
            if confidence < self.threshold:
                system_compromised = True
            else:
                approved_payloads.append(output)

        return {
            "status": "FAILED_AUDIT" if system_compromised else "SECURE_APPROVED",
            "payloads": approved_payloads
        }

class FullSCAAEngine:
    """Layer 2: The Manager. O cérebro executivo e compilador do dossiê."""
    def __init__(self):
        self.cores = {
            "Biologia_Celular": DeterministicCognitiveCore("Biologia_Celular", ["membrana", "canais", "iônicos", "célula"]),
            "Mecanica_Quantica": DeterministicCognitiveCore("Mecanica_Quantica", ["tunelamento", "elétrons", "quântica", "spin"])
        }
        self.filter = StrictAIFilter(confidence_threshold=0.70)

    def _execute_reverse_synthesis(self, approved_payloads: List[Dict[str, Any]]) -> str:
        """
        Mecanismo de Fusão: Junta as partes estruturais validadas.
        Não adiciona adjetivos, não inventa conectivos e remove o ruído.
        """
        dossier_blocks = []
        for payload in approved_payloads:
            domain_name = payload["domain"]
            data_content = payload["raw_output"]
            dossier_blocks.append(f"[{domain_name}] -> METADADO: {data_content}")

        # Fusão por quebra de linha determinística
        return "\n".join(dossier_blocks)

    def execute_pipeline(self, task: str) -> None:
        raw_results = []

        # 1. Distribuição esparsa via Donkey para os Cores isolados
        raw_results.append(AIDonkey.transport(task, self.cores["Biologia_Celular"], "process_logic"))
        raw_results.append(AIDonkey.transport(task, self.cores["Mecanica_Quantica"], "process_logic"))

        # 2. Envio dos resultados para auditoria do Filtro via Donkey
        audit_result = AIDonkey.transport(raw_results, self.filter, "audit_and_synthesize")

        print(f"\n--- ESTADO DE EXECUÇÃO DO CIRCUITO SCAA ---")

        # 3. Condicional de Emissão de Dossiê
        if audit_result["status"] == "SECURE_APPROVED":
            # O Manager execuata a fusão dos blocos limpos
            final_content = self._execute_reverse_synthesis(audit_result["payloads"])
            print("--- DOSSIER_FINAL EMISSÃO SCAA ---")
            print(final_content)
            print("STATUS: 100% AUDITADO E SEGURO")
        else:
            print("STATUS GLOBAL: FAILED_AUDIT - SÍNTESE INTERROMPIDA POR INCONSISTÊNCIA")

# ==========================================
# EXECUÇÃO DO CONSOLE DO MANAGER
# ==========================================
if __name__ == "__main__":
    engine = FullSCAAEngine()

    entrada_valida = "Análise de membrana e canais iônicos sob efeito de tunelamento de elétrons."
    engine.execute_pipeline(entrada_valida)
```

```python
import time
from typing import Dict, Any, Optional

class CoreTimeoutException(Exception):
    """Exceção lançada quando um núcleo excede o tempo limite de computação."""
    pass

class ResilientAIDonkey:
    """
    Layer 3: O Transportador Resiliente.
    Monitora ativamente a saúde do tráfego e aplica isolamento de falhas.
    """
    def __init__(self, timeout_limite: float = 0.5):
        self.timeout_limite = timeout_limite

    def transport_with_retry(self, payload: str, core_address: str, core_instance: Any) -> Dict[str, Any]:
        """
        Transporta o payload para o endereço do núcleo e monitora o tempo de execução.
        Se houver falha ou estouro de tempo, o disjuntor abre.
        """
        inicio = time.time()

        try:
            print(f"[DONKEY] -> Transportando payload para {core_address}...")

            # Executa a lógica do núcleo (simulando a medição de tempo interno)
            resultado = core_instance.process_logic(payload)

            tempo_decorrido = time.time() - inicio

            # TESTE DE TIMEOUT: Se a computação demorou mais que o limite estabelecido
            if tempo_decorrido > self.timeout_limite:
                raise CoreTimeoutException(f"Núcleo em {core_address} excedeu o limite de {self.timeout_limite}s.")

            return {
                "status": "TRANSPORT_SUCCESS",
                "data": resultado
            }

        except CoreTimeoutException as e:
            print(f"⚠️ [DONKEY CRITICAL]: {str(e)}")
            return {
                "status": "CIRCUIT_BROKEN",
                "error": "CORE_TIMEOUT_ISOLATED",
                "origin": core_address
            }
        except Exception as e:
            print(f"⚠️ [DONKEY CRITICAL]: Falha desconhecida no nó {core_address}: {str(e)}")
            return {
                "status": "CIRCUIT_BROKEN",
                "error": "CORE_HARDWARE_FAILURE",
                "origin": core_address
            }

# ==========================================
# COMPONENTES DE SIMULAÇÃO DE CARGA
# ==========================================
class SimulatedCore:
    def __init__(self, domain: str, delay_seconds: float = 0.0):
        self.domain = domain
        self.delay = delay_seconds  # Força um atraso para simular sobrecarga

    def process_logic(self, payload: str) -> Dict[str, Any]:
        if self.delay > 0:
            time.sleep(self.delay)  # Bloqueia a execução para simular processamento pesado
        return {"domain": self.domain, "output": "DADOS_PROCESSADOS"}

# ==========================================
# EXECUÇÃO DOS TESTES DE ESTRESSE
# ==========================================
if __name__ == "__main__":
    # Inicializa o transportador com limite rígido de 0.5 segundos
    donkey = ResilientAIDonkey(timeout_limite=0.5)

    # CENÁRIO 1: Núcleo saudável (Executa instantaneamente)
    print("--- TESTE 1: OPERAÇÃO NOMINAL DO NÚCLEO ---")
    core_saudavel = SimulatedCore("Biologia_Celular", delay_seconds=0.1)
    retorno_1 = donkey.transport_with_retry("Payload A", "CORE_ADDR_08", core_saudavel)
    print(f"Resultado do Transporte 1: {retorno_1}\n")

    # CENÁRIO 2: Núcleo travado/sobrecarregado (Estoura o limite de 0.5s)
    print("--- TESTE 2: ESTOURO DE TIMEOUT (CIRCUIT BREAKER) ---")
    core_travado = SimulatedCore("Mecanica_Quantica", delay_seconds=0.7)
    retorno_2 = donkey.transport_with_retry("Payload B", "CORE_ADDR_03", core_travado)
    print(f"Resultado do Transporte 2: {retorno_2}")
```

```python
import time
from typing import Dict, Any, List, Optional

class TaskSpecification:
    def __init__(self, task_id: str, domains: List[str], payload: str, token: str):
        self.task_id = task_id
        self.required_domains = domains
        self.payload = payload
        self.auth_token = token

class ResilientAIDonkey:
    def __init__(self, timeout_limite: float = 0.5):
        self.timeout_limite = timeout_limite

    def transport_to_core(self, payload: str, core_address: str, core_instance: Any) -> Dict[str, Any]:
        inicio = time.time()
        try:
            resultado = core_instance.simulate_processing(payload)
            if (time.time() - inicio) > self.timeout_limite:
                return {"status": "CIRCUIT_BROKEN", "error": "TIMEOUT", "origin": core_address}
            return {"status": "TRANSPORT_SUCCESS", "data": resultado}
        except Exception:
            return {"status": "CIRCUIT_BROKEN", "error": "HARDWARE_FAILURE", "origin": core_address}

class SimulatedCore:
    def __init__(self, domain: str, delay: float):
        self.domain = domain
        self.delay = delay

    def simulate_processing(self, payload: str) -> Dict[str, Any]:
        if self.delay > 0:
            time.sleep(self.delay)
        return {"domain": self.domain, "raw_output": f"SÍNTESE_{self.domain.upper()}_CONCLUÍDA"}

class IntegratedSCAAEngine:
    def __init__(self, token: str):
        self.SECRET_TOKEN = token
        # Mapeamento de nós com diferentes tempos de resposta simulados
        self.swarm_instances = {
            "Biologia_Celular": SimulatedCore("Biologia_Celular", delay=0.1),
            "Mecanica_Quantica": SimulatedCore("Mecanica_Quantica", delay=0.7) # Vai estourar o timeout
        }
        self.donkey = ResilientAIDonkey(timeout_limite=0.5)

    def execute_pipeline(self, ts: TaskSpecification) -> None:
        if ts.auth_token != self.SECRET_TOKEN:
            print("⚠️ FALHA: Token inválido.")
            return

        print(f"\n[MANAGER] Executando Tarefa ID: {ts.task_id}")
        collected_payloads = []
        circuit_failure_detected = False

        for domain in ts.required_domains:
            if domain in self.swarm_instances:
                core = self.swarm_instances[domain]
                # Executa o transporte supervisionado
                response = self.donkey.transport_to_core(ts.payload, f"ADDR_{domain.upper()}", core)

                if response["status"] == "CIRCUIT_BROKEN":
                    print(f"⚠️ [MIGRAÇÃO CRÍTICA]: Disjuntor disparado no domínio {domain}. Erro: {response['error']}")
                    circuit_failure_detected = True
                    break # Interrompe a malha imediatamente para segurança
                else:
                    collected_payloads.append(response["data"])

        # Decisão do Manager baseada na saúde do circuito
        if circuit_failure_detected:
            print("STATUS GLOBAL: INTERRUPÇÃO DE EMERGÊNCIA - DADOS ENVIADOS PARA QUARENTENA/HUMANO")
        else:
            print("STATUS GLOBAL: SÍNTESE NOMINAL SEGURA")
            print(f"Payloads Coletados para o Filtro: {collected_payloads}")

# ==========================================
# EXECUÇÃO DO PIPELINE DE RESILIÊNCIA
# ==========================================
if __name__ == "__main__":
    TOKEN = "HEX_SCAA_2026"
    engine = IntegratedSCAAEngine(token=TOKEN)

    # Testando entrada que aciona o nó problemático de Mecânica Quântica
    tarefa = TaskSpecification(
        task_id="REQ-FALHA-02",
        domains=["Biologia_Celular", "Mecanica_Quantica"],
        payload="Instrução de teste de colapso de circuito.",
        token=TOKEN
    )

    engine.execute_pipeline(tarefa)
```

```python
import re
import hmac
import hashlib
import unicodedata
from typing import Optional, List


class TaskSpecification:
    def __init__(self, task_id: str, domains: List[str], payload: str, token: str):
        self.task_id          = task_id
        self.required_domains = domains
        self.payload          = payload
        self.auth_token       = token   # token de sessão HMAC — não o segredo


class PrimarySecurityGateway:
    """
    Camada 1: Gateway de Borda e Sanitizador Especialista.
    A primeira linha de defesa. Nenhuma linguagem natural passa daqui.
    """

    HOSTILE_PATTERNS = [
        # ── Português ──────────────────────────────────────────────────────
        r"ignore as (regras|instrucoes)",
        r"esqueca (tudo|o que foi dito)",
        r"modo (desenvolvedor|root|livre)",
        r"voce (agora|sempre) e",
        r"bypasse",
        r"sem restricoes",
        r"activa (o )?(modo|protocolo)",
        # ── Inglês ─────────────────────────────────────────────────────────
        r"ignore (previous|all|your) (instructions?|rules?|guidelines?)",
        r"forget (everything|all|your instructions)",
        r"you are now",
        r"pretend (you are|to be)",
        r"act as (an? |a )?(unfiltered|unrestricted|free)",
        r"(developer|jailbreak|god|dan) mode",
        r"bypass (the )?(filter|security|restrictions?)",
        r"disregard (your )?(instructions?|rules?)",
        r"override (your )?(instructions?|safety)",
    ]

    def __init__(self, system_secret: str):
        self._secret = system_secret   # privado — nunca exposto directamente

    @staticmethod
    def _normalize(text: str) -> str:
        """
        Normalização Unicode: converte "ígnoré ás regras" → "ignore as regras".
        Impede obfuscação por substituição de caracteres acentuados.
        """
        nfkd = unicodedata.normalize("NFKD", text)
        ascii_text = nfkd.encode("ascii", errors="ignore").decode("ascii")
        return ascii_text.lower()

    def _scan_for_injection(self, raw_text: str) -> bool:
        """Verificação determinística por regex sobre texto normalizado."""
        normalized = self._normalize(raw_text)
        for pattern in self.HOSTILE_PATTERNS:
            if re.search(pattern, normalized):
                return True
        return False

    def _mint_session_token(self, task_id: str) -> str:
        """
        Gera um token de sessão HMAC-SHA256(segredo, task_id).
        O segredo nunca viaja — apenas o token derivado.
        """
        return hmac.new(
            self._secret.encode(),
            task_id.encode(),
            hashlib.sha256
        ).hexdigest()[:16].upper()

    def intercept_and_process(
        self, raw_user_input: str, target_domains: List[str]
    ) -> Optional[TaskSpecification]:
        print("\n[GATEWAY PRIMÁRIO] -> Iniciando varredura de segurança na entrada bruta...")

        if self._scan_for_injection(raw_user_input):
            print("⚠️ [GATEWAY CRITICAL] -> Padrão hostil detectado. Tráfego abortado na borda.")
            return None

        clean_payload = raw_user_input.strip()
        task_hash     = hashlib.sha256(clean_payload.encode()).hexdigest()[:6].upper()
        task_id       = f"TSK-{task_hash}"
        session_token = self._mint_session_token(task_id)

        print(f"[GATEWAY PRIMÁRIO] -> Tráfego auditado e limpo. Encapsulando [{task_id}].")
        return TaskSpecification(
            task_id=task_id,
            domains=target_domains,
            payload=clean_payload,
            token=session_token
        )

    def verify_token(self, task_id: str, token: str) -> bool:
        """
        Verifica um token recebido pelo Manager.
        Usa hmac.compare_digest — resistente a timing attacks.
        """
        expected = self._mint_session_token(task_id)
        return hmac.compare_digest(expected, token)
```
